```python
## Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Load the Dataset
df = pd.read_csv("data/netflix_titles.csv", encoding="latin1")
df.head()
```

```python
## Data Cleaning and Preprocessing

df.isnull().sum()
```

```python
df['country'] = df['country'].fillna("Unknown")
df['rating'] = df['rating'].fillna("Not Rated")
```

```python
## Movies vs TV Shows Distribution 
type_count = df['type'].value_counts()
type_count
```

```python
type_count = df['type'].value_counts()

type_count.plot(
    kind='bar',
    color=['#1f77b4', '#ff7f0e']  # blue & orange
)

plt.title("Distribution of Movies and TV Shows on OTT Platforms")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.show()
```

```python
## Content Trend Over the Years 
yearly_content = df['release_year'].value_counts().sort_index()
yearly_content.tail(10) 
```

```python
yearly_content = df['release_year'].value_counts().sort_index()

yearly_content.plot(
    kind='line',
    color='#2ca02c',   # green
    linewidth=2
)

plt.title("OTT Content Release Trend Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Releases")

plt.show()
```

```python
## Top Content Producing Countries

top_countries = df['country'].value_counts().head(5)
top_countries
```

```python
top_countries = df['country'].value_counts().head(5)

top_countries.plot(
    kind='bar',
    color='#9467bd'   # purple
)

plt.title("Top 5 Countries Producing OTT Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
## Most Common Content Ratings
rating_count = df['rating'].value_counts().head(5)
rating_count
```

```python
rating_count = df['rating'].value_counts().head(5)

rating_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']
)

plt.title("Distribution of Top Content Ratings")
plt.ylabel("")
plt.show()
```

```python
## Numerical Analysis using NumPy

movie_years = df[df['type'] == 'Movie']['release_year']
average_year = np.mean(movie_years)

print("Average release year of movies:", int(average_year))
```

```python
## Final Insights
print("Total Records:", len(df))
print("Total Movies:", len(df[df['type'] == 'Movie']))
print("Total TV Shows:", len(df[df['type'] == 'TV Show']))
```
